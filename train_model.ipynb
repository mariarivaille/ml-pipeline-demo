```python
import pandas as pd
import numpy as np
import pickle
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline

# 1. Загружаем данные
# Используем seaborn для простоты, либо можно pd.read_csv('titanic.csv')
import seaborn as sns
df = sns.load_dataset('titanic')

# 2. Отбираем признаки и целевую переменную
# Очистим данные от явных мусорных колонок для демо
features = ['pclass', 'sex', 'age', 'sibsp', 'parch', 'fare', 'embarked']
target = 'survived'

X = df[features]
y = df[target]

# 3. Настраиваем предобработку
# Числовые признаки
numeric_features = ['age', 'sibsp', 'parch', 'fare']
numeric_transformer = SimpleImputer(strategy='median')

# Категориальные признаки
categorical_features = ['pclass', 'sex', 'embarked']
categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore'))
])

# Объединяем предобработку
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numeric_features),
        ('cat', categorical_transformer, categorical_features)
    ])

# 4. Создаем полный пайплайн (Предобработка + Модель)
clf = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('model', LogisticRegression(max_iter=1000))
])

# 5. Обучаем
clf.fit(X, y)

# 6. Проверяем точность (для себя)
print(f"Accuracy on training data: {clf.score(X, y):.2f}")

# 7. Сохраняем модель
with open('model.pkl', 'wb') as f:
    pickle.dump(clf, f)

print("Model saved to model.pkl")
```
